```python
import sqlite3
import sys
 
def inspect_db(db_path):
    conn = sqlite3.connect(db_path)
    cur = conn.cursor()
 
    tables = [r[0] for r in cur.execute(
        "SELECT name FROM sqlite_master WHERE type='table';"
    ).fetchall()]
 
    print(f"DB: {db_path}")
    print(f"Tables: {tables}\n")
 
    for t in tables:
        print(f"--- {t} ---")
        cols = cur.execute(f"PRAGMA table_info({t});").fetchall()
        for cid, name, ctype, notnull, dflt, pk in cols:
            print(f"  {name:30s} {ctype:10s} {'PK' if pk else ''}")
        count = cur.execute(f"SELECT COUNT(*) FROM {t};").fetchone()[0]
        print(f"  rows: {count}\n")
 
    conn.close()
 

inspect_db('/mnt/research/IceCube/lownutau/gemini_moreMC_sqlite/nutau_train/nutau_gemini_ftp_65TeV_lowerbound_skimmed_chunk_022860.021990_0000.db_train.db')
```
